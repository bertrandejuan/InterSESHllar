```python
# from the tutorial https://www.youtube.com/watch?v=lyT4BLEboyQ&t=40s

import earthaccess
import xarray
import hvplot.xarray
```

```python
auth = earthaccess.login()
# login to earth access, create account here: https://urs.earthdata.nasa.gov/profile
```

```python
query = earthaccess.search_datasets(
    keyword =   'EMITL2BMIN'
)

len(query)
```

```python
for r in query[:5]:
    long_name = r.get_umm('EntryTitle')
    short_name = r.get_umm('ShortName')
    description = r.get_umm('Abstract')
    doi = r.get_umm('DOI')
    print(long_name, short_name, description, doi, sep = '\n')
    print( )
```

```python
lon, lat = (45.131878, 59.844488)
radius = 1000000. #m

results = earthaccess.search_data(
    doi = '10.5067/EMIT/EMITL2BMIN.002',
    circle = (lon, lat, radius),
    temporal = ('2026-10-01', '2026-10-30'),
)
len(results)
```

```python
files = earthaccess.download(results[:1])
```

```python
ds = xarray.open_dataset(files[1], group = 'mineral_metadata')
ds
```

```python
ds['h_li'].hvplot(kind = 'scatter', x = 'latitude', s =2)
```
